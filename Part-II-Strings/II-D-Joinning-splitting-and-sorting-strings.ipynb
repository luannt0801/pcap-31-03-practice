# Python offers two ways of sorting.

`sorted()` dùng được với string và trả về list các ký tự đã sắp xếp.

The first one is a function named sorted.

This function takes a list as its argument and returns a new list with the elements sorted.

The original list remains unchanged.

`sort()` là method của list
The second option is to use sort not sorted.

So it is a bit different because it changes the original list.

No new list is created.

The syntax is therefore also a bit different.

sorted(string)
     ↓
list mới

string.sort()
     ↓
❌ không tồn tại

list.sort()
     ↓
sắp xếp chính list đó

```python
names = ['Adam', 'Kate', 'Barbara', 'Donna']

names_sorted = sorted(['Adam', 'Kate', 'Barbara', 'Donna'])

print(names_sorted)
```

```python
names.sort()
print(names)
```

```python

def get_longest_word(input_string):
    list_split = input_string.split()
    sort_list = list_split.sort()
    return sort_list[len(input_string)-1]
```

```python
def get_longest_word(input_string):
    words = input_string.replace('.', ' ').replace(',', ' ').split()
    temp_max_word = ''
 
    for word in words:
      if len(word) > len(temp_max_word):
        temp_max_word = word
    
    return temp_max_word
```
