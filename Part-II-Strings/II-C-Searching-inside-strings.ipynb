Hàm `.index()` này để tìm vị trí mà ký tự đang thuộc về trong chuỗi string

````python
'ilovetravellingaroundtheworld'.index('a')
````

````python
'ilovetravellingaroundtheworld'[7]
````

- `index()` trả về giá trị đầu tiên
- `find()` tìm oder tương ứng

|           | Tìm thấy     | Không tìm thấy |
| --------- | ------------ | -------------- |
| `find()`  | vị trí index | `-1`           |
| `index()` | vị trí index | `ValueError`   |


Ví dụ thực tế:
```python
if text.find("Java") != -1:
    print("Found")
```
Trong khi với index() thường dùng try/except:
```python
try:
    position = text.index("Java")
except ValueError:
    print("Not found")
```

````python
text = "xin chao bo con nha may"
print(text.index('bo'))
print(text.find('bo'))
````

`rfind()` = reverse find → tìm vị trí xuất hiện cuối cùng của substring.

`find()` → tìm từ trái sang phải, trả về lần xuất hiện đầu tiên

`rfind()` → tìm từ phải sang trái, trả về lần xuất hiện cuối cùng

````python
text = "abc abc abc"

print(text.find("abc"))    # 0
print(text.rfind("abc"))   # 8

print(text.index("abc"))   # 0
print(text.rindex("abc"))  # 8
````

````python
print('work work work'.find('work'))

print('work work work'.rfind('work'))

print('work work work'.find('work', 2)) # Bắt đầu tìm từ index 2:

print('work work work'.rfind('work', 2)) # tìm từ phải sang trái, nhưng chỉ trong phạm vi từ index 2 trở đi → 10.

print('work work work'.find('work', 2, 5)) # Tìm "work" trong: index 2 → trước index 5

print('work work work'.rfind('work', 2, 5))
````
