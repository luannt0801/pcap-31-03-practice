# Propagating Exceptions

````python
def get_day(user_info):
    day = int(input('What is the day of your bday?'))
    user_info.append(day)


def get_month(user_info):
    month = int(input('What is the month (1-12) of your bday?'))
    user_info.append(month)


def get_year(user_info):
    year = int(input('What is the year of your bday?'))
    user_info.append(year)


def get_user_bday(user_info):
    try:
        get_day(user_info)
        get_month(user_info)
        get_year(user_info)

        print('So your bday is', user_info)
    except ValueError:
        print("You entered incorrected data, bye!")

print('Hi, I will collect some info about your bday!')
user_info = []
get_user_bday(user_info)
````

Chạy code chỉ khi phần try khkoong xảy ra exception

```python
try:

except SomeError:

else:
    # Chỉ chạy nếu KHÔNG có exception

finally:
    # Luôn chạy, bất kể có exception hay không
```

````python
try:
    value = int(input('enter: '))
    print(1/value)
except:
    print('something went wrong!')
else:
    print('dcm')
````

raise tạo exception → except xử lý nó → else chỉ chạy khi không có exception nào xảy ra trong try.

## 1. Exception là object

Khi Python gặp:

```python
raise ValueError("Something went wrong")
```

Python sẽ tạo ra một **exception object** thuộc class `ValueError`.

Có thể hiểu:

```text
ValueError  → class
     ↓
ValueError("Something went wrong") → object
```

Vì exception là object nên bạn có thể truy cập thông tin của nó.

---

## 2. `except ... as e` lấy exception object

Ví dụ:

```python
try:
    1 / 0
except Exception as e:
    print(e)
```

Ở đây:

```python
as e
```

cho phép bạn **đặt tên cho exception object** là `e`.

Bạn có thể sử dụng:

```python
e
e.args
str(e)
```

---

## 3. `print(e)` cho thông tin lỗi

Ví dụ:

```python
try:
    5 < "a"
except Exception as e:
    print(e)
```

Có thể nhận được:

```text
'<' not supported between instances of 'int' and 'str'
```

Tại sao `print(e)` không in địa chỉ object?

Vì `Exception` đã định nghĩa `__str__()` để chuyển exception object thành **thông báo lỗi có ý nghĩa**.

---

## 4. Exception có hierarchy

Exception trong Python được tổ chức theo **inheritance hierarchy**.

Phần quan trọng:

```text
BaseException
      │
      ├── SystemExit
      ├── KeyboardInterrupt
      └── Exception
            │
            ├── ValueError
            ├── TypeError
            ├── ArithmeticError
            ├── LookupError
            └── ...
```

Vì vậy:

```python
issubclass(ValueError, Exception)
```

→ `True`

Và:

```python
issubclass(Exception, BaseException)
```

→ `True`

Điều này giải thích tại sao:

```python
except Exception:
```

có thể bắt được `ValueError`, `TypeError`, v.v.

---

## 5. `Exception.__subclasses__()`

Có thể xem các class **trực tiếp kế thừa** từ `Exception`:

```python
print(Exception.__subclasses__())
```

Lưu ý: đây chỉ là **direct subclasses**, không phải toàn bộ exception hierarchy.

---

## 6. `e.args` là gì? ⭐

Đây là phần rất đáng nhớ.

Mỗi exception có thuộc tính:

```python
e.args
```

Đây là một **tuple chứa các arguments được truyền vào constructor của exception**.

Ví dụ:

```python
try:
    raise Exception()
except Exception as e:
    print(e.args)
```

→

```text
()
```

Không truyền argument nào.

---

### Truyền 1 argument

```python
try:
    raise Exception("I do not like it")
except Exception as e:
    print(e.args)
```

→

```text
('I do not like it',)
```

Đây là tuple **1 phần tử**, nên có dấu `,`.

---

### Truyền nhiều arguments

```python
try:
    raise Exception("I do not like it", "at all")
except Exception as e:
    print(e.args)
```

→

```text
('I do not like it', 'at all')
```

Vì:

```python
Exception(a, b, c)
```

→

```python
e.args == (a, b, c)
```

---

## 7. Python cũng tự thêm `args` khi phát sinh lỗi

Bạn không nhất thiết phải tự `raise`.

Ví dụ:

```python
try:
    1 / 0
except Exception as e:
    print(e.args)
```

Python tự tạo `ZeroDivisionError` và cung cấp thông tin:

```text
('division by zero',)
```

Tương tự:

```python
try:
    5 < "a"
except Exception as e:
    print(e.args)
```

Python đưa thông tin mô tả lỗi vào `args`.

# Cheat sheet cần nhớ

| Khái niệm                    | Ý nghĩa                                |
| ---------------------------- | -------------------------------------- |
| `Exception`                  | Một class cơ sở cho nhiều exception    |
| `raise`                      | Chủ động tạo/phát sinh exception       |
| `except ... as e`            | Lấy exception object vào biến `e`      |
| `print(e)`                   | In thông báo lỗi                       |
| `e.args`                     | Tuple chứa arguments của exception     |
| `Exception.__subclasses__()` | Các subclass trực tiếp của `Exception` |
| `issubclass(A, B)`           | Kiểm tra A có kế thừa B không          |

### Ví dụ tổng hợp

```python
try:
    raise ValueError("Invalid age", 20)

except Exception as e:
    print(type(e))
    print(e)
    print(e.args)
```

Kết quả về ý nghĩa:

```text
<class 'ValueError'>
Invalid age
('Invalid age', 20)
```

**3 thứ nên nhớ nhất từ bài này:**

> **Exception là object → `as e` lấy object → `e.args` chứa các arguments được truyền vào exception.**
