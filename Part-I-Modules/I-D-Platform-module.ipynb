

The platform module can tell you a little bit about your Python version, the operating

system, and the hardware you're using.



## The platform module allows us to access some basic information about the underlying layers the Python environment, the operating system and the hardware.

```python
import platform
```

```python
platform.platform()
```

```python
platform.platform(aliased= False, terse=False)
```

```python
platform.platform(aliased= True, terse=True)
```

```python
platform.machine()
```

```python
platform.processor()
```

```python
platform.system()
```

```python
platform.python_implementation()
```

```python
platform.architecture()
```

`version()` trả về system release version

```python
platform.python_version()
```

`python_version_tuple()` trả về python version

```python
platform.python_version_tuple()
```

platform.python_version_tuple() vs. platform.version()
In the previous video, we've discussed platform.python_version_tuple(), which returns your Python version as a 3-element tuple, for example:

('3', '8', '12')

There's also another function that looks similar: platform.version(). This function, in turn, returns your system's release version as a single string, for example:

#24~20.04.1-Ubuntu SMP Mon Sep 12 06:14:01 UTC 2022

Keep in mind not to confuse these two functions for the PCAP exam!
