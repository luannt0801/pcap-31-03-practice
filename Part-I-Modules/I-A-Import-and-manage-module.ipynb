## Import và quản lý các module
Module giúp chia code ra thành các phần gồm các file nhằm mục đích dễ quản lý và mở rộng code base.



Tương tác với module sử dụng các lệnh đưới dây:
1. Lệnh `dir`: Lệnh này show các func, object trong 1 module
ví dụ `dir(numpy)` sẽ trả về toàn bộ các class, object, sub-module trong `numpy`
2. Thư viện `sys`: là các lệnh tương tác với trực tiếp với hệ điều hành, quản lý các process đang chạy.
	- Lệnh `sys.exit(f"message")` sẽ tắt process hiện tại đang chạy khi gặp dòng lệnh này.
	- Lệnh `sys.`
3. Có thể sử dụng `import` + tên module + `*` để import toàn bộ những gì có trong module.
   Nhưng nếu dùng `from sys import *` thì cần phải ghi nhớ các lệnh này ở trong module nào? vì khi đó khi gọi các object trong module, không cần phải khai báo module đó. 
   trường hợp này nên dùng `import sys as s`
4. Đổi hàm biến, khi import bằng tên khác có thể sử dụng: `from sys import exit as bye_bye`
