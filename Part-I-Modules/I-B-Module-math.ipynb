## Module `math`

Đây là 1 module căn bản trong python giúp ta sử lý các hàm với toán học.

```python
import math
```

```python
for name in dir(math):
    print(name, end='\t')
```

### Các hàm toán học về làm tròn số hoặc lấy phần nguyên số thập phân

- `ceil()`: làm tròn lên
- `floor()`: làm tròn xuống
- `trunc()`: chỉ lấy phần nguyên

```python
# ceil()
print(math.ceil(3.4))
print(math.ceil(3.5))
print(math.ceil(3.6))
print(math.ceil(-3.6))
```

```python
# floor()
print(math.floor(3.4))
print(math.floor(3.5))
print(math.floor(3.6))
print(math.floor(-5.4))
```

```python
# trunc()
print(math.trunc(3.4))
print(math.trunc(3.5))
print(math.trunc(3.6))
print(math.trunc(-2.6))
```

Ta có thể sử dụng hàm `round()` để linh động khi làm tròn lên/xuống

```python
# round()
import numpy as np
print(np.round(3.4))
print(np.round(3.5))
print(np.round(3.6))
print(np.round(-5.4))
```

Tính giai thừa sử dụng hàm `factorial()

```python
print(math.factorial(3))
print(math.factorial(4))
```

```python
# sqrt()
print(math.sqrt(16))
print(math.sqrt(25))
```

```python
# hyport()
# dùng để tính cạnh huyền tam giác vuông

print(math.hypot(3, 4))
```
