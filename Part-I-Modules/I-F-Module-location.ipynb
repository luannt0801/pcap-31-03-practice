# Sử dụng thư viện sys.path để quản lý location của các module



• Xác định vị trí tìm kiếm module: Khi bạn gọi import my_module, Python sẽ quét qua lần lượt các đường dẫn có trong sys.path để tìm file my_module.py.

• Thứ tự ưu tiên: Python tìm kiếm theo thứ tự các phần tử trong danh sách sys.path. Phần tử đầu tiên (thường là thư mục chứa script đang chạy hoặc thư mục hiện tại) sẽ được ưu tiên cao nhất.

• Tùy biến đường dẫn động: Vì sys.path là một list thông thường, bạn có thể thêm (append, insert) hoặc xóa các đường dẫn thư mục tùy ý ngay trong mã nguồn Python khi chương trình đang chạy. Điều này rất hữu ích khi bạn muốn import các module nằm ở một thư mục khác biệt ngoài cấu trúc dự án thông thường.

```python
from sys import path
```

Khi trình thông dịch Python khởi động, sys.path được điền giá trị từ các nguồn sau:
1. Thư mục chứa đoạn mã lệnh (script) đang được thực thi (hoặc thư mục làm việc hiện tại nếu chạy giao diện tương tác).
2. Biến môi trường PYTHONPATH (nếu có được thiết lập trên hệ điều hành).
3. Các thư mục mặc định của thư viện chuẩn Python và các gói bên thứ ba (thư mục site-packages).

```python
print(path)
```

---

Sử dụng `sys.path.append('folder chứa module')` để add module, lúc đó chỉ cần gọi `import module_da_append_path`

```python
import sys
sys.path.append('F://blabla_module')
import blabla
```

# Khái niệm `namespace`


Namespace (không gian tên) trong Python là một hệ thống ánh xạ (mapping) từ tên biến, hàm hoặc lớp tới các đối tượng trong bộ nhớ, giúp đảm bảo mỗi tên gọi là duy nhất và tránh xung đột
