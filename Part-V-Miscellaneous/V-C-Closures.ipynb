# A closure is a technique for implementing lexically scoped name binding in a language with first-class functions.

````python
def greet(text):
    def print_greet():
        print(text)
    return print_greet

say_hello = greet('hello')
say_hello()
````

````python
say_hello
````

````python
say_hello is greet
````

# Một function bên trong có thể "nhớ" các biến của function bên ngoài, ngay cả sau khi function bên ngoài đã chạy xong.

x là biến của function bên ngoài.

`multiply()` sử dụng x, mặc dù x không được truyền vào `multiply()`.

                 ┌─────────────┐
multiply_5 ─────→│ multiply()  │
                 └──────┬──────┘
                        │
                        ↓
                    x = 5

Tại sao không viết đơn giản hơn?

Bạn có thể viết:
```python
def multiply(x, y):
    return x * y

multiply(5, 10)
multiply(12, 10)
```
Nhưng closure hữu ích khi bạn muốn tạo ra những function có behavior đã được cấu hình sẵn.

`Closure = function + environment (các biến bên ngoài mà function đó "nhớ").`

````python
def make_multiply_closure(x):
    def multiply(y):
        return x * y

    return multiply


multiply_5 = make_multiply_closure(5)
multiply_12 = make_multiply_closure(12)

print(multiply_5(10))
print(multiply_5(20))

print(multiply_12(10))
print(multiply_12(20))
````
