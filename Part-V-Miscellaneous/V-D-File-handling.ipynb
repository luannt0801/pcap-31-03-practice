```python
def count_occurences(file_name, word):
    with open(file_name, "r") as file:
        text = file.read().lower()

    words = text.split()
    count = 0

    for w in words:
        w = w.strip(".,!?;:'\"()[]{}")
        if w == word.lower():
            count += 1

    return count

print(count_occurences(file_name='dcm.txt', word='the'))
```

```python
def count_occurences(file_name, word):
    count = 0
    try:
        stream = open(file_name, 'r')
        for line in stream:
            words = line.replace('.', ' ').replace(',', ' ').split()
            for i in words:
                if(i.lower() == word.lower()):
                    count = count + 1
    except Exception:
        print('something went wrong')
    finally:
        stream.close()
    return count
```
