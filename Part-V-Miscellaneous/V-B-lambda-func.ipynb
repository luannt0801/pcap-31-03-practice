# `map()` and `filter()` with lambdas

`map()` trong Python dùng để áp dụng một function lên từng phần tử của một iterable (list, tuple, ...).

```python
lambda_func = lambda i : i*2
initial_list = [1,2,3,4]
```

```python
map(lambda_func, initial_list)
```

```python
map_result = map(lambda_func, initial_list)
print(next(map_result))
print(next(map_result))
print(next(map_result))
print(next(map_result))
print(next(map_result))
```

```python
map_result = map_result = map(lambda_func, initial_list)
for element in map_result:
    print(element, end='-')
```

`map()` → biến đổi từng phần tử

`filter()` → chọn/giữ lại những phần tử thỏa điều kiện

```python
print(list(filter(lambda i: i %2 ==0, [2,4,6,8,9,11,13])))
```
