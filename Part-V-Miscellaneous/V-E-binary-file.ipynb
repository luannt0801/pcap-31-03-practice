```python
data = bytearray(100)
data[0]=100
data[1]=201

try:
    stream = open('file.bin', 'wb')
    stream.write(data)
    stream.close()
except Exception as e:
    print(e)
```

# stream error

```python
try:
    stream = open('nonexistent.txt')
    stream.close()
except Exception as e:
    print(e.errno)
```

```python
from os import strerror
try:
    stream = open('nonexistent.txt')
    stream.close()
except Exception as e:
    print(strerror(e.errno))
```
