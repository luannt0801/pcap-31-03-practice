# SpaceX Falcon 9 – Data Collection with the SpaceX REST API
Collect past launch data from the SpaceX REST API, enrich it with booster, launch site, payload and core information, keep only Falcon 9 launches and clean missing values.

```python
import requests
import pandas as pd
import numpy as np
import datetime

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
```

## Helper functions that call the SpaceX API for each ID

```python
def getBoosterVersion(data):
    for x in data['rocket']:
        if x:
            response = requests.get("https://api.spacexdata.com/v4/rockets/" + str(x), timeout=20).json()
            BoosterVersion.append(response['name'])

def getLaunchSite(data):
    for x in data['launchpad']:
        if x:
            response = requests.get("https://api.spacexdata.com/v4/launchpads/" + str(x), timeout=20).json()
            Longitude.append(response['longitude'])
            Latitude.append(response['latitude'])
            LaunchSite.append(response['name'])

def getPayloadData(data):
    for load in data['payloads']:
        if load:
            response = requests.get("https://api.spacexdata.com/v4/payloads/" + load, timeout=20).json()
            PayloadMass.append(response['mass_kg'])
            Orbit.append(response['orbit'])

def getCoreData(data):
    for core in data['cores']:
        if core['core'] != None:
            response = requests.get("https://api.spacexdata.com/v4/cores/" + core['core'], timeout=20).json()
            Block.append(response['block'])
            ReusedCount.append(response['reuse_count'])
            Serial.append(response['serial'])
        else:
            Block.append(None)
            ReusedCount.append(None)
            Serial.append(None)
        Outcome.append(str(core['landing_success']) + ' ' + str(core['landing_type']))
        Flights.append(core['flight'])
        GridFins.append(core['gridfins'])
        Reused.append(core['reused'])
        Legs.append(core['legs'])
        LandingPad.append(core['landpad'])
```

## Request the launch data
The live endpoint is `https://api.spacexdata.com/v4/launches/past`. For consistent results the course provides a static JSON snapshot of the same response.

```python
spacex_url = "https://api.spacexdata.com/v4/launches/past"
try:
    live = requests.get(spacex_url, timeout=20)
    print("Live API status code:", live.status_code)
except Exception as e:
    print("Live API not reachable:", e)

static_json_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/API_call_spacex_api.json'
response = requests.get(static_json_url)
print("Static snapshot status code:", response.status_code)
```

```python
# Decode the response content as JSON and turn it into a Pandas dataframe
data = pd.json_normalize(response.json())
data.head()
```

```python
print("First static_fire_date_utc:", data['static_fire_date_utc'].iloc[0])
```

## Keep the relevant features and clean the rows

```python
data = data[['rocket', 'payloads', 'launchpad', 'cores', 'flight_number', 'date_utc']]

# Remove rows with multiple cores (Falcon Heavy) and multiple payloads
data = data[data['cores'].map(len) == 1]
data = data[data['payloads'].map(len) == 1]

# Extract the single value from the lists
data['cores'] = data['cores'].map(lambda x: x[0])
data['payloads'] = data['payloads'].map(lambda x: x[0])

# Convert date_utc to a date and restrict the launch dates
data['date'] = pd.to_datetime(data['date_utc']).dt.date
data = data[data['date'] <= datetime.date(2020, 11, 13)]
data.shape
```

## Enrich each launch with details from the API

```python
BoosterVersion, PayloadMass, Orbit, LaunchSite, Outcome = [], [], [], [], []
Flights, GridFins, Reused, Legs, LandingPad = [], [], [], [], []
Block, ReusedCount, Serial, Longitude, Latitude = [], [], [], [], []

api_ok = True
try:
    getBoosterVersion(data)
    getLaunchSite(data)
    getPayloadData(data)
    getCoreData(data)
except Exception as e:
    api_ok = False
    print("SpaceX API detail endpoints unavailable:", type(e).__name__)

if api_ok:
    launch_dict = {'FlightNumber': list(data['flight_number']), 'Date': list(data['date']),
                   'BoosterVersion': BoosterVersion, 'PayloadMass': PayloadMass, 'Orbit': Orbit,
                   'LaunchSite': LaunchSite, 'Outcome': Outcome, 'Flights': Flights,
                   'GridFins': GridFins, 'Reused': Reused, 'Legs': Legs, 'LandingPad': LandingPad,
                   'Block': Block, 'ReusedCount': ReusedCount, 'Serial': Serial,
                   'Longitude': Longitude, 'Latitude': Latitude}
    launch_df = pd.DataFrame(launch_dict)
else:
    # Fallback: the course's published result of exactly these API calls
    print("Using the IBM course snapshot of the enriched API data instead.")
    launch_df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv')
    launch_df['BoosterVersion'] = 'Falcon 9'
launch_df.head()
```

## Filter the dataframe to only include Falcon 9 launches

```python
data_falcon9 = launch_df[launch_df['BoosterVersion'] != 'Falcon 1'].copy()
data_falcon9.loc[:, 'FlightNumber'] = list(range(1, data_falcon9.shape[0] + 1))
print("Falcon 9 launches:", len(data_falcon9))
```

## Data wrangling – missing values

```python
data_falcon9.isnull().sum()
```

```python
# Replace missing PayloadMass values with the mean
payload_mean = data_falcon9['PayloadMass'].mean()
data_falcon9['PayloadMass'] = data_falcon9['PayloadMass'].replace(np.nan, payload_mean)
print("Missing LandingPad values:", data_falcon9['LandingPad'].isnull().sum())
data_falcon9.isnull().sum()
```

```python
data_falcon9.to_csv('dataset_part_1.csv', index=False)
```
