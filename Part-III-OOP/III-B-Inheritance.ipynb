# Inheriting properties

```python
class Vehicle:
    def __init__(self, speed):
        self.speed = speed
        
class LandVehicle(Vehicle):
    pass

class Car(LandVehicle):
    pass
```

```python
my_car = Car(speed=120)
print(my_car.__dict__)
```

### sss

```python
class Vehicle:
    def __init__(self, speed):
        self.speed = speed
        
class LandVehicle(Vehicle):
    def __init__(self, wheel_count):
        self.wheel_count = wheel_count

class Car(LandVehicle):
    pass
```

```python
my_car = Car(speed=120)
print(my_car.__dict__)
```

```python
my_car = Car(wheel_count=120)
print(my_car.__dict__)
```

## sss2

```python
class Vehicle:
    def __init__(self, speed):
        self.speed = speed
        
class LandVehicle(Vehicle):
    def __init__(self, speed, wheel_count):
        Vehicle.__init__(self, speed) # = super.__init__(self, speed)
        self.wheel_count = wheel_count

class Car(LandVehicle):
    pass
```

```python
my_car = Car(wheel_count=120)
print(my_car.__dict__)
```

```python
my_car = Car(speed=120)
print(my_car.__dict__)
```

```python
my_car = Car(speed=120, wheel_count=4)
print(my_car.__dict__)
```
