Để access 1 private attribute của 1 class thì call `_nameclass__privateattributes` để gọi

```python
class Dog:
    def __init__(self, name, age):
        self.__name = name
        self.age = age
```

```python
my_dog = Dog("aloi", 10)
print(my_dog.age)
print(my_dog.__name)
```

```python
import random

class Card:
    def __init__(self, suit , value):
        self.suit  = suit 
        self.value = value
        
    def present(self):
        return f"{self.value} of {self.suit}"
    
class Deck:
    suits = ['hearts', 'diamonds', 'clubs', 'spades']
    values = ['Ace', '2', '3', '4', '5', '6', '7', '8', '9', '10', 'Jack', 'Queen', 'King']
    def __init__(self):
        self.cards: list[Card] = self.__create_cards()
        
    def __create_cards(self):
        cards = []
        for suit in self.suits:
            for value in self.values:
                cards.append(Card(suit, value))
        return cards
    
    def shuffle(self):
        random.shuffle(self.cards)
    
    def deal(self):
        if not self.cards:
            return None
        else:
            card_dealed = self.cards.pop(random.randrange(len(self.cards)))
            return card_dealed
    
    def count_remaining(self):
        return len(self.cards)
    
    def get_remaining(self):
        return [card.present() for card in self.cards]
```
